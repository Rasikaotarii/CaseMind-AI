````python
import pandas as pd
import numpy as np
import matplotlib as mtp
import seaborn as sb
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
````



```
# This is formatted as code
```
create dataset



````python
Data={"size":[700,900,1000,1000,1200,1500,1500,1700,1800,1800,2000,2200],
      "BHK":[1,1,1,2,2,2,3,3,3,4,4,4],
      "price":[1800000,2000000,2200000,2500000,2750000,3200000,3500000,4000000,4200000,4500000,4800000,5000000]}
````

````python
Data
````

````python
df=pd.DataFrame(Data)
````

````python
df
````

````python
df.corr()
````

visualize using seaborn

````python
sb.pairplot(df)
````

dependant and independant variable

````python
X=df[["size","BHK"]]
y=df["price"]
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.02, random_state=42)
````

model training

````python
model=LinearRegression()
````

````python
model.fit(X_train,y_train)
````

scatter plot


````python
print("Intercept:",model.intercept_)
print("coef:",model.coef_)
````

make prediction


````python
y_pred=model.predict(X_test)
````

````python
print("predicted:",y_pred)
print("actual:",y_test.values)
````

model evaluate

````python
mse=mean_squared_error(y_test,y_pred)
````

````python
mse
````

````python
print("r2 score",r2_score)
````

predict the price of flat


````python
A=[[1800,2]]
print("predicted price:",model.predict(A))
````
